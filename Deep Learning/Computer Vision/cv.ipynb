```python
import cv2
import numpy as np
car_cascade = cv2.CascadeClassifier('haarcascade_car.xml')
image = cv2.imread('c:/Users/vigne/Downloads/images.jpg')
if image is None:
    print("Error:Could not load image. Please check the file path.")
else:
    image = cv2.resize(image, (400,200))
    cv2.imshow('Original image', image)
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    cars = car_cascade.detectMultiScale(gray_image, 1.1, 1)
    for (x, y, w, h) in cars:
        cv2.rectangle(image, (x, y), (x+w, y+h), (0, 255, 0), 2)
    cv2.imshow('Detected Cars', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
video_path = 'c:/Users/vigne/Downloads/16727473-hd_1080_1920_30fps.mp4'
cap = cv2.VideoCapture(video_path)
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    # Process the frame here
    cv2.imshow('Frame', frame)
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print("Error: Could not open webcam.")
while True:
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break
    cv2.imshow('Webcam', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import os
output_dir = 'saved_frames'
os.makedirs(output_dir, exist_ok=True)
cap = cv2.VideoCapture('c:/Users/vigne/Downloads/16727473-hd_1080_1920_30fps.mp4')
if not cap.isOpened():
    print("Error opening video file")
    exit()
frame_number = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame_filename = os.path.join(output_dir, f'frame_{frame_number:04d}.jpg')
    cv2.imwrite(frame_filename, frame)
    frame_number += 1
cap.release()
print("All frames saved successfully")
```
