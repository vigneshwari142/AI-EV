```python
import pandas as pd
import numpy as np
data = pd.read_csv(r'C:\Users\vigne\OneDrive\Desktop\AI-EV VIGNESHWARI\Unsupervised learning\Clustering\environmental_socioeconomic.csv')
data.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)
print(pd.DataFrame(data_scaled, columns=data.columns).head())
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
plt.figure(figsize=(10,7))
dendrogram = sch.dendrogram(sch.linkage(data_scaled, method='ward'))
plt.title('Dendrogram for Hierarichcal Clustering')
plt.xlabel('Index')
plt.ylabel('Euclidean Distance')
plt.show()
```

```python
hierarchical_clustering = AgglomerativeClustering(n_clusters=7, metric='euclidean', linkage='ward')
data['cluster'] = hierarchical_clustering.fit_predict(data_scaled)
print(data.head())
```

```python
from sklearn.metrics import silhouette_score
sil_score = silhouette_score(data_scaled, data['cluster'])
print(f'Silhouette Score:{sil_score}')
```

```python
import seaborn as sns
plt.figure(figsize=(8,6))
sns.scatterplot(x='co2_emissions', y='waste_production', hue='cluster', data=data, palette='viridis', s=100, alpha=0.7, edgecolor='k')
plt.title('hierarchical_clustering of Environmental and Socioeconomic Factors')
plt.xlabel('CO2 Emissions')
plt.ylabel('Waste Production')
plt.legend(title='Cluster',bbox_to_anchor=(1.05,1), loc='upper left')
plt.show()
```
